```python
#class

class student:
    def __init__(self,name,cource):
        #constructor
        self.name = name
        self.cource = cource

    def display(self):
        print(f"{self.name} is enrolled in {self.cource}")
```

```python
s1 = student("alice","datascience")
s2 = student("bob","machine learning")
```

```python
s1.display()
s2.display()
```
